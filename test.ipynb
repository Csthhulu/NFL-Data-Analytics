```python
import pandas as pd
import BDAAData as BDAAData
import glob
import matplotlib.pyplot as plt


file_paths = glob.glob(r'C:\BDAA Folder\BDAAData\*.csv')
df = pd.concat([pd.read_csv(file_path).round(2) for file_path in file_paths], ignore_index=True)
# df = pd.read_csv(r'C:\BDAA Folder\BDAAData\2026_plays.csv')
df.head()         # first 5 rows
df.shape          # (rows, columns)
df.info()         # types + non-null counts
df.describe()     # stats for number columns
df.dropna(inplace=True)  # remove rows with missing values
df.isna().sum()   # missing values per column

# top = df.head(10)      # show the first 10 terms
# bottom = df.tail(10)   # show the last 10 terms
# concate = pd.concat([top, bottom])  # concatenate top and bottom slices
# concate.head(20)  # show the first 20 rows of the concatenated DataFrame

results = df['IsScoringDrive'].value_counts().reindex([0, 1], fill_value=0)

results.index = ['no', 'yes']

results.plot(kind='bar')
plt.title('Distribution of Scoring Drives')
plt.xlabel('Drive Number')
plt.ylabel('Scoring Drives')
plt.show()
```
